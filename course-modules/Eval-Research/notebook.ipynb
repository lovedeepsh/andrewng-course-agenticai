# Eval & Research — Component-level evaluation of the research step

Instead of evaluating a whole research pipeline (search → draft → reflect), this
lab evaluates just **one component**: the web-search/research step. The check is
**objective** with a **per-example ground truth** — do the returned URLs come from
a predefined set of preferred domains?

Research + tools are in `src/` (provided). The evaluation function in
`src/evaluation.py` is yours to write — it's the point of this lab.


```python
%load_ext autoreload
%autoreload 2

import json
from dotenv import load_dotenv

from src import config
from src.research import find_references
from src.evaluation import evaluate_tavily_results
from src.rendering import print_html

load_dotenv()
```

## 1. The research step

`find_references` lets the LLM use arxiv / tavily / wikipedia tools to gather
sources. Run it on the black-hole task.

```python
research_result = find_references(config.RESEARCH_TASK)
print_html(research_result, title="Research Function Output")
```

## 2. The evaluation (you write `evaluate_tavily_results`)

Extract the URLs, compare each domain against `config.TOP_DOMAINS`, compute the
preferred ratio, and return a PASS/FAIL flag plus a Markdown report. Then this
cell runs it.

```python
print_html(json.dumps(sorted(config.TOP_DOMAINS)[:4], indent=2), title="Sample Preferred Domains")

flag, report = evaluate_tavily_results(config.TOP_DOMAINS, research_result, min_ratio=config.MIN_RATIO)
print_html(report, title="Evaluation Summary")
print("Result:", "PASS" if flag else "FAIL")
```

## 3. Try it yourself

Change the topic, the threshold, or the preferred-domain set, then re-run.

```python
topic = "recent developments in black hole science"
min_ratio = 0.4
top_domains = {"wikipedia.org", "nature.com", "science.org", "arxiv.org",
               "nasa.gov", "mit.edu", "stanford.edu", "harvard.edu"}

research_output = find_references(f"Find 2-3 key papers and reliable overviews about {topic}.")
print_html(research_output, title=f"Research Results on {topic}")

flag, eval_md = evaluate_tavily_results(top_domains, research_output, min_ratio=min_ratio)
print_html(eval_md, title="Evaluation Summary")
print("Result:", "PASS" if flag else "FAIL")
```
